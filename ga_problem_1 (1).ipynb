```python
import random
import numpy as np

# ==========================================
# Problem 1: Maximize f(x) = x^2 (0 <= x <= 31)
# ==========================================

def decode(chromosome):
    return int(chromosome, 2)

def fitness_function(x):
    return x**2

# STEP 1 
population_notes = ['01100', '11001', '00101', '10011']

x_vals_notes = [decode(c) for c in population_notes]
fits_notes = [fitness_function(x) for x in x_vals_notes]
total_fit = sum(fits_notes)
avg_fit = total_fit / len(population_notes)

print(f"{'No.':<5} | {'Pop':<7} | {'X':<5} | {'f(x)=X^2':<10} | {'Prob':<10} | {'Expected':<10}")
print("-" * 60)

probabilities = []
expected_counts = []
for i in range(len(population_notes)):
    prob = fits_notes[i] / total_fit
    exp_count = fits_notes[i] / avg_fit
    probabilities.append(prob)
    expected_counts.append(exp_count)
    print(f"{i+1:<5} | {population_notes[i]:<7} | {x_vals_notes[i]:<5} | {fits_notes[i]:<10} | {prob:<10.3f} | {exp_count:<10.3f}")

actual_counts = [round(exp) for exp in expected_counts]
print(f"\nActual Counts (Rounded): {actual_counts}")

mating_pool = []
for i, count in enumerate(actual_counts):
    for _ in range(count):
        mating_pool.append(population_notes[i])
print(f"Mating Pool: {mating_pool}")
print("\n" + "="*50 + "\n")

# --- END TO END GENETIC ALGORITHM ---
def select_parents(population, fitnesses):
    total_fit = sum(fitnesses)
    if total_fit == 0:
        return random.choices(population, k=2)
    probs = [f / total_fit for f in fitnesses]
    return random.choices(population, weights=probs, k=2)

def crossover(p1, p2):
    point = random.randint(1, len(p1)-1)
    c1 = p1[:point] + p2[point:]
    c2 = p2[:point] + p1[point:]
    return c1, c2

def mutate(chromosome, mutation_rate=0.05):
    mutated = ""
    for bit in chromosome:
        if random.random() < mutation_rate:
            mutated += '0' if bit == '1' else '1'
        else:
            mutated += bit
    return mutated

def run_genetic_algorithm(generations=20, pop_size=4, mutation_rate=0.05):
    print("--- STARTING END-TO-END EVOLUTION ---")
    population = ['01100', '11001', '00101', '10011']
    while len(population) < pop_size:
        population.append(''.join(random.choice('01') for _ in range(5)))
        
    best_chrom = None
    best_fit = -1

    print("Generation | Best f(x) | Avg f(x) | Best Chromosome | Best X")
    print("-" * 65)

    for gen in range(generations):
        x_values = [decode(c) for c in population]
        fitnesses = [fitness_function(x) for x in x_values]
        
        max_f = max(fitnesses)
        avg_f = sum(fitnesses) / pop_size
        best_idx = fitnesses.index(max_f)
        
        if max_f > best_fit:
            best_fit = max_f
            best_chrom = population[best_idx]
            
        print(f"{gen:<10} | {max_f:<9} | {avg_f:<8.2f} | {population[best_idx]:<15} | {x_values[best_idx]:<6}")
        
        next_gen = [population[best_idx]]
        while len(next_gen) < pop_size:
            p1, p2 = select_parents(population, fitnesses)
            c1, c2 = crossover(p1, p2)
            next_gen.append(mutate(c1, mutation_rate))
            if len(next_gen) < pop_size:
                next_gen.append(mutate(c2, mutation_rate))
                
        population = next_gen

    print("\n--- EVOLUTION COMPLETE ---")
    print(f"Overall Best Chromosome: {best_chrom}")
    print(f"Overall Best X Value: {decode(best_chrom)}")
    print(f"Overall Best Fitness f(x): {best_fit}")

if __name__ == '__main__':
    run_genetic_algorithm(generations=15, pop_size=6, mutation_rate=0.1)
```
