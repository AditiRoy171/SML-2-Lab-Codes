```python
import random
import numpy as np

# ==========================================
# Problem 2: Maximize f(x) = -x^2 + 2x (0 <= x <= 2)
# ==========================================

def decode_continuous(chromosome, min_val=0, max_val=2, bits=5):
    decimal = int(chromosome, 2)
    return min_val + ((max_val - min_val) / ((2**bits) - 1)) * decimal

def fitness_function(x):
    return -(x**2) + 2*x

#STEP 1
population_p2 = ['11010', '00111', '10110', '00101']
print(f"{'Pop':<7} | {'Decimal':<7} | {'Encoded X':<10} | {'f(x)':<10}")
print("-" * 45)

fitness_p2 = []
for chrom in population_p2:
    dec = int(chrom, 2)
    x_val = decode_continuous(chrom, 0, 2, 5)
    fit = fitness_function(x_val)
    fitness_p2.append(fit)
    print(f"{chrom:<7} | {dec:<7} | {x_val:<10.3f} | {fit:<10.3f}")

total_fit_p2 = sum(fitness_p2)
print(f"\nTotal Fitness: {total_fit_p2:.3f}")

# Using the random numbers provided
random_numbers = [0.4, 0.15, 0.7, 0.9]
probabilities_p2 = [f / total_fit_p2 for f in fitness_p2]
cumulative_prob = np.cumsum(probabilities_p2)

print(f"\nRoulette Wheel Selection (Using your random numbers: {random_numbers}):")
selected_parents = []
for r in random_numbers:
    for i, cum_p in enumerate(cumulative_prob):
        if r <= cum_p:
            selected_parents.append(population_p2[i])
            print(f"Random number {r} selects parent {population_p2[i]}")
            break

print(f"\nSelected Mating Pool: {selected_parents}")
print("\n" + "="*50 + "\n")

# --- END TO END GENETIC ALGORITHM ---
def select_parents_roulette(population, fitnesses):
    total_fit = sum(fitnesses)
    if total_fit == 0:
        return random.choices(population, k=2)
    probs = [f / total_fit for f in fitnesses]
    return random.choices(population, weights=probs, k=2)

def crossover(p1, p2):
    point = random.randint(1, len(p1)-1)
    c1 = p1[:point] + p2[point:]
    c2 = p2[:point] + p1[point:]
    return c1, c2

def mutate(chromosome, mutation_rate=0.05):
    mutated = ""
    for bit in chromosome:
        if random.random() < mutation_rate:
            mutated += '0' if bit == '1' else '1'
        else:
            mutated += bit
    return mutated

def run_genetic_algorithm(generations=20, pop_size=4, mutation_rate=0.05):
    print("--- STARTING END-TO-END EVOLUTION ---")
    population = ['11010', '00111', '10110', '00101']
    
    while len(population) < pop_size:
        population.append(''.join(random.choice('01') for _ in range(5)))
        
    best_chrom = None
    best_fit = -float('inf')

    print("Generation | Best f(x) | Avg f(x) | Best Chromosome | Best X")
    print("-" * 65)

    for gen in range(generations):
        x_values = [decode_continuous(c, 0, 2, 5) for c in population]
        fitnesses = [fitness_function(x) for x in x_values]
        
        max_f = max(fitnesses)
        avg_f = sum(fitnesses) / pop_size
        best_idx = fitnesses.index(max_f)
        
        if max_f > best_fit:
            best_fit = max_f
            best_chrom = population[best_idx]
            
        print(f"{gen:<10} | {max_f:<9.3f} | {avg_f:<8.3f} | {population[best_idx]:<15} | {x_values[best_idx]:<6.3f}")
        
        next_gen = [population[best_idx]]
        while len(next_gen) < pop_size:
            p1, p2 = select_parents_roulette(population, fitnesses)
            c1, c2 = crossover(p1, p2)
            next_gen.append(mutate(c1, mutation_rate))
            if len(next_gen) < pop_size:
                next_gen.append(mutate(c2, mutation_rate))
                
        population = next_gen

    print("\n--- EVOLUTION COMPLETE ---")
    print(f"Overall Best Chromosome: {best_chrom}")
    print(f"Overall Best X Value: {decode_continuous(best_chrom, 0, 2, 5):.3f}")
    print(f"Overall Best Fitness f(x): {best_fit:.3f}")

if __name__ == '__main__':
    run_genetic_algorithm(generations=15, pop_size=6, mutation_rate=0.1)
```
